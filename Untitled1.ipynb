```python
import inspect

def get_init_attributes(obj):
    """Get only attributes assigned in __init__"""
    # Get the parameters that __init__ expects
    init_signature = inspect.signature(obj.__class__.__init__)
    init_params = list(init_signature.parameters.keys())[1:]  # Remove 'self'
    
    # Get attributes that match __init__ parameters
    init_attrs = {}
    for param in init_params:
        if hasattr(obj, param):
            try:
                init_attrs[param] = getattr(obj, param)
            except:
                init_attrs[param] = "<Unable to read>"
    
    return init_attrs

class Person:
    def __init__(self, name, age, city="Unknown"):
        self.name = name
        self.age = age
        self.city = city
        self.computed_attr = f"{name} from {city}"  # This is computed, not input

person = Person("Alice", 25, "New York")
person.dynamic_attr = "This should not appear"

print("Only __init__ assigned attributes:")
for attr, value in get_init_attributes(person).items():
    print(f"  {attr}: {value}")

get_init_attributes(person)
```

```python
import json
import numpy as np
from sklearn.base import BaseEstimator

class SklearnEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, BaseEstimator):
            return self.serialize_estimator(obj)
        elif isinstance(obj, np.ndarray):
            return obj.tolist()
        elif isinstance(obj, np.integer):
            return int(obj)
        elif isinstance(obj, np.floating):
            return float(obj)
        elif hasattr(obj, '__dict__'):
            return obj.__dict__
        return super().default(obj)
    
    def serialize_estimator(self, estimator):
        """Serialize a scikit-learn estimator"""
        serialized = {
            '__class__': estimator.__class__.__name__,
            '__module__': estimator.__module__,
            'params': estimator.get_params() if hasattr(estimator, 'get_params') else {}
        }
        
        # Add fitted attributes (those ending with _)
        fitted_params = {}
        for attr in dir(estimator):
            if attr.endswith('_') and not attr.startswith('_'):
                try:
                    value = getattr(estimator, attr)
                    if hasattr(value, 'tolist'):  # numpy arrays
                        fitted_params[attr] = value.tolist()
                    else:
                        fitted_params[attr] = value
                except:
                    continue
        
        if fitted_params:
            serialized['fitted_attributes'] = fitted_params
            
        return serialized

```
